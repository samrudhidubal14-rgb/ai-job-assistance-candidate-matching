```python
# ============================================================
# AI-BASED INTELLIGENT JOB ASSISTANCE & CANDIDATE MATCHING
# Complete End-to-End Google Colab Project
# ============================================================

# -----------------------------
# 1. INSTALL / IMPORT LIBRARIES
# -----------------------------
!pip -q install scikit-learn pandas numpy matplotlib seaborn joblib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

warnings.filterwarnings("ignore")

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print("=" * 70)
print("AI-BASED INTELLIGENT JOB ASSISTANCE & CANDIDATE MATCHING SYSTEM")
print("=" * 70)


# -----------------------------
# 2. UPLOAD DATASET
# -----------------------------
print("\nSTEP 1: Upload your CSV dataset")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\nDataset successfully loaded!")
print("Dataset Shape:", df.shape)

display(df.head())


# -----------------------------
# 3. BASIC DATA INFORMATION
# -----------------------------
print("\n" + "=" * 70)
print("STEP 2: DATASET INFORMATION")
print("=" * 70)

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate Rows:", df.duplicated().sum())


# -----------------------------
# 4. IDENTIFY TARGET
# -----------------------------
target = "Target_Category"

if target not in df.columns:
    raise ValueError(
        "Target_Category column was not found in the dataset."
    )

print("\nTarget Variable:", target)

print("\nTarget Distribution:")
display(df[target].value_counts())


# -----------------------------
# 5. REMOVE DUPLICATES
# -----------------------------
df = df.drop_duplicates().reset_index(drop=True)

print("\nDataset after removing duplicates:", df.shape)


# -----------------------------
# 6. EXPLORATORY DATA ANALYSIS
# -----------------------------
print("\n" + "=" * 70)
print("STEP 3: EXPLORATORY DATA ANALYSIS")
print("=" * 70)

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x=target,
    order=df[target].value_counts().index
)

plt.title("Candidate Suitability Distribution")
plt.xlabel("Candidate Category")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# -----------------------------
# 7. NUMERICAL FEATURES
# -----------------------------
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

if target in numerical_columns:
    numerical_columns.remove(target)

if len(numerical_columns) > 0:

    print("\nNumerical Features:")
    print(numerical_columns)

    df[numerical_columns].hist(
        figsize=(16, 12),
        bins=25
    )

    plt.suptitle(
        "Numerical Feature Distributions",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()


# -----------------------------
# 8. CORRELATION MATRIX
# -----------------------------
if len(numerical_columns) > 1:

    plt.figure(figsize=(14, 10))

    correlation = df[numerical_columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm"
    )

    plt.title("Correlation Matrix")
    plt.tight_layout()
    plt.show()


# -----------------------------
# 9. PREPARE FEATURES
# -----------------------------
print("\n" + "=" * 70)
print("STEP 4: DATA PREPROCESSING")
print("=" * 70)

# Remove target
X = df.drop(columns=[target])

y = df[target]

# Remove ID-type columns
id_columns = []

for col in X.columns:

    if (
        "id" in col.lower()
        or "candidate_id" in col.lower()
    ):
        id_columns.append(col)

if id_columns:

    print("Removing identifier columns:", id_columns)

    X = X.drop(columns=id_columns)


# -----------------------------
# 10. IDENTIFY DATA TYPES
# -----------------------------
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("\nNumerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)


# -----------------------------
# 11. PREPROCESSING PIPELINE
# -----------------------------
numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)


# -----------------------------
# 12. TRAIN TEST SPLIT
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)


# -----------------------------
# 13. RANDOM FOREST MODEL
# -----------------------------
print("\n" + "=" * 70)
print("STEP 5: MODEL TRAINING")
print("=" * 70)

model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                random_state=42,
                n_jobs=-1,
                class_weight="balanced"
            )
        )
    ]
)

print("\nTraining Random Forest...")

model.fit(
    X_train,
    y_train
)

print("Model training completed successfully!")


# -----------------------------
# 14. PREDICTION
# -----------------------------
y_pred = model.predict(X_test)


# -----------------------------
# 15. MODEL EVALUATION
# -----------------------------
print("\n" + "=" * 70)
print("STEP 6: MODEL EVALUATION")
print("=" * 70)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    "\nModel Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# -----------------------------
# 16. CONFUSION MATRIX
# -----------------------------
cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

plt.tight_layout()
plt.show()


# -----------------------------
# 17. FEATURE IMPORTANCE
# -----------------------------
print("\n" + "=" * 70)
print("STEP 7: FEATURE IMPORTANCE")
print("=" * 70)

trained_preprocessor = model.named_steps[
    "preprocessor"
]

classifier = model.named_steps[
    "classifier"
]

feature_names = (
    trained_preprocessor
    .get_feature_names_out()
)

importance = classifier.feature_importances_

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

feature_importance = (
    feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
)

print("\nTop 20 Important Features:")

display(
    feature_importance.head(20)
)


# -----------------------------
# 18. FEATURE IMPORTANCE GRAPH
# -----------------------------
top_features = feature_importance.head(20)

plt.figure(figsize=(10, 8))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 20 Features Influencing Candidate Suitability"
)

plt.tight_layout()
plt.show()


# -----------------------------
# 19. CANDIDATE PREDICTION FUNCTION
# -----------------------------
print("\n" + "=" * 70)
print("STEP 8: CANDIDATE JOB MATCHING")
print("=" * 70)


def predict_candidate(candidate_data):

    candidate_df = pd.DataFrame(
        [candidate_data]
    )

    prediction = model.predict(
        candidate_df
    )[0]

    print("\nCandidate Prediction:")
    print("--------------------------------")

    print(
        "Suitability Category:",
        prediction
    )

    # Probability if supported
    try:

        probabilities = (
            model.predict_proba(
                candidate_df
            )[0]
        )

        probability_table = pd.DataFrame({

            "Category":
            model.classes_,

            "Probability":
            probabilities

        })

        probability_table[
            "Probability"
        ] = (
            probability_table[
                "Probability"
            ] * 100
        ).round(2)

        print("\nPrediction Probabilities:")

        display(
            probability_table.sort_values(
                by="Probability",
                ascending=False
            )

        )

    except Exception:
        pass

    return prediction


# -----------------------------
# 20. AUTOMATIC SAMPLE CANDIDATE
# -----------------------------
print("\nCreating a sample candidate...")

sample_candidate = {}

for column in X.columns:

    if column in numerical_features:

        # Use median value for numerical columns
        sample_candidate[column] = (
            X[column].median()
        )

    elif column in categorical_features:

        # Use most common category
        sample_candidate[column] = (
            X[column]
            .mode()[0]
        )


prediction = predict_candidate(
    sample_candidate
)


# -----------------------------
# 21. SAVE MODEL
# -----------------------------
print("\n" + "=" * 70)
print("STEP 9: SAVE TRAINED MODEL")
print("=" * 70)

model_file = (
    "AI_Job_Assistance_Candidate_Matching_Model.pkl"
)

joblib.dump(
    model,
    model_file
)

print(
    "\nModel saved as:",
    model_file
)


# -----------------------------
# 22. DOWNLOAD MODEL
# -----------------------------
files.download(
    model_file
)


# -----------------------------
# 23. FINAL SUMMARY
# -----------------------------
print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nProject:",
    "AI-Based Intelligent Job Assistance & Candidate Matching System"
)

print(
    "Dataset Size:",
    df.shape
)

print(
    "Model:",
    "Random Forest Classifier"
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Target:",
    target
)

print(
    "\nThe trained model can now classify candidates"
    " based on the available candidate/job features."
)

print("\nNext Step:")
print(
    "Connect this model to a Streamlit web application"
    " and deploy it using GitHub + Streamlit Cloud."
)

print("=" * 70)
```
